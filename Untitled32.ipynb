```python
df["fl_date"] = pd.to_datetime(df["fl_date"])

df["op_unique_carrier"] = df["op_unique_carrier"].astype("category")
df["origin"] = df["origin"].astype("category")
df["origin_city_name"] = df["origin_city_name"].astype("category")
df["origin_state_nm"] = df["origin_state_nm"].astype("category")
df["dest"] = df["dest"].astype("category")
df["dest_city_name"] = df["dest_city_name"].astype("category")
df["dest_state_nm"] = df["dest_state_nm"].astype("category")
df["cancellation_code"] = df["cancellation_code"].astype("category")

print(df.dtypes)
```

```python
df = df.dropna(subset=["op_carrier_fl_num"])
```

```python
print(
    df.groupby("cancelled")[["dep_time", "dep_delay", "arr_delay"]]
      .apply(lambda x: x.isnull().sum())
)
```

```python
print(
    df[(df["cancelled"] == 0) & (df["arr_delay"].isnull())][
        ["dep_time", "dep_delay", "arr_time", "arr_delay", "diverted"]
    ].head(20)
)
```

```python
import numpy as np

df["flight_status"] = np.select(
    [
        df["cancelled"] == 1,
        df["diverted"] == 1
    ],
    [
        "Cancelled",
        "Diverted"
    ],
    default="Completed"
)
```

```python
print(df["flight_status"].value_counts())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
df["dep_delay"] = pd.to_numeric(df["dep_delay"], errors="coerce")
```

```python
df = df.dropna(subset=["origin", "dest"])
```

```python
print(df.isnull().sum())
```

```python
import numpy as np

df["delay_status"] = np.where(
    df["dep_delay"] > 0,
    "Delayed",
    "On Time"
)
```

```python
df["delay_category"] = np.select(
    [
        df["dep_delay"] <= 0,
        df["dep_delay"] <= 15,
        df["dep_delay"] <= 60,
        df["dep_delay"] > 60
    ],
    [
        "On Time",
        "Small Delay",
        "Medium Delay",
        "Severe Delay"
    ],
    default="Unknown"
)
```

```python
total_flights = len(df)

print(total_flights)
```

```python
delayed = (df["dep_delay"] > 0).sum()

print(delayed)
```

```python
average_delay = df["dep_delay"].mean()

print(average_delay)
```

```python
airline_flights = df["op_unique_carrier"].value_counts()

print(airline_flights)
```

```python
airline_delay = df.groupby(
    "op_unique_carrier",
    observed=True
)["dep_delay"].mean()

print(airline_delay)
```

```python
destination_flights = df["dest"].value_counts()

print(destination_flights.head(10))
```

```python
df["route"] = (
    df["origin"].astype(str)
    + " → "
    + df["dest"].astype(str)
)
```

```python
top_routes = df["route"].value_counts().head(10)

print(top_routes)
```

```python
delay_causes = df[
    [
        "carrier_delay",
        "nas_delay",
        "weather_delay",
        "security_delay",
        "late_aircraft_delay"
    ]
].sum()

print(delay_causes)
```

```python
df["dep_hour"]=df["crs_dep_time"]//100
df["arr_hour"]=df["crs_arr_time"]//100
df["is_delayed"]=np.where(df["dep_delay"]>0,1,0)
df["is_severe_delay"]=np.where(df["dep_delay"]>60,1,0)
df["delay_minutes"]=df["dep_delay"].clip(lower=0)
df["distance_category"]=np.select([df["distance"]<500,df["distance"]<1000,df["distance"]<2000,df["distance"]>=2000],["Short","Medium","Long","Very Long"],default="Unknown")
df["flight_duration_category"]=np.select([df["actual_elapsed_time"]<120,df["actual_elapsed_time"]<240,df["actual_elapsed_time"]>=240],["Short","Medium","Long"],default="Unknown")
delay_cols=["carrier_delay","weather_delay","nas_delay","security_delay","late_aircraft_delay"]
df["main_delay_cause"]=df[delay_cols].idxmax(axis=1)
```

```python
print(df.info())
```

```python
# 1. Check data
print("Shape:", df.shape)
df.info()

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Check duplicate rows
print("\nDuplicates:", df.duplicated().sum())

# 4. Basic statistics
print("\nBasic Statistics:")
print(df.describe())

# 5. Flight status
print("\nFlight Status:")
print(df["flight_status"].value_counts())

# 6. Delayed flights
print("\nDelayed Flights:")
print(df["is_delayed"].value_counts())

# 7. Delay causes
print("\nDelay Causes:")
print(df["main_delay_cause"].value_counts())

# 8. Flights by month
print("\nFlights by Month:")
print(df.groupby("month", observed=True).size())

# 9. Average delay by airline
print("\nAverage Delay by Airline:")
print(df.groupby("op_unique_carrier", observed=True)["delay_minutes"].mean().sort_values(ascending=False))

# 10. Average delay by month
print("\nAverage Delay by Month:")
print(df.groupby("month", observed=True)["delay_minutes"].mean())

# 11. Flights by distance category
print("\nDistance Category:")
print(df["distance_category"].value_counts())

# 12. Flights by duration category
print("\nFlight Duration Category:")
print(df["flight_duration_category"].value_counts())

# 13. Top origin airports
print("\nTop 10 Origin Airports:")
print(df["origin"].value_counts().head(10))

# 14. Top destination airports
print("\nTop 10 Destination Airports:")
print(df["dest"].value_counts().head(10))

# 15. Top routes
print("\nTop 10 Routes:")
print(df["route"].value_counts().head(10))
```

```python
# 1. Fix airline groupby warning
airline_delay = df.groupby("op_unique_carrier", observed=True)["delay_minutes"].mean().sort_values(ascending=False)

# 2. Check top 10 routes
print("Top 10 Routes:")
print(df["route"].value_counts().head(10))

# 3. Fix main delay cause
delay_cols = ["carrier_delay","weather_delay","nas_delay","security_delay","late_aircraft_delay"]
df["main_delay_cause"] = np.where(df[delay_cols].sum(axis=1)==0,"No Delay",df[delay_cols].idxmax(axis=1).str.replace("_delay","",regex=False))

# 4. Check delay causes
print("Delay Causes:")
print(df["main_delay_cause"].value_counts())

# 5. Create final columns for Power BI
final_columns = [
    "year","month","day_of_month","day_of_week","fl_date",
    "op_unique_carrier","op_carrier_fl_num",
    "origin","origin_city_name","origin_state_nm",
    "dest","dest_city_name","dest_state_nm",
    "crs_dep_time","dep_time","dep_delay","taxi_out","wheels_off",
    "wheels_on","taxi_in","crs_arr_time","arr_time","arr_delay",
    "cancelled","cancellation_code","diverted",
    "crs_elapsed_time","actual_elapsed_time","air_time","distance",
    "carrier_delay","weather_delay","nas_delay","security_delay",
    "late_aircraft_delay","flight_status","delay_status",
    "delay_category","route",
    "dep_hour","arr_hour","is_delayed","is_severe_delay",
    "delay_minutes","distance_category","flight_duration_category",
    "main_delay_cause"
]

# 6. Create final dataframe
df_final = df[final_columns].copy()

# 7. Check final data
print("Final Shape:", df_final.shape)
print(df_final.head())

# 8. Check missing values
print("Missing Values:")
print(df_final.isnull().sum())

# 9. Check duplicates
print("Duplicate Rows:", df_final.duplicated().sum())

# 10. Save final dataset
df_final.to_csv("flight_data_final.csv", index=False)

print("Final dataset saved successfully!")
```
